```python
import torch
from transformers import AutoModelForCausalLM
from PIL import Image

# Ładowanie modelu (na Colabie zadziała bez kwantyzacji)
model_id = "vikhyatk/moondream2"
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    revision="2025-06-21",
    trust_remote_code=True,
    torch_dtype=torch.float16,
    device_map="cuda"
)

# Analiza zdjęcia
image = Image.open("auto.jpg").convert("RGB")

with torch.no_grad():
    caption = model.caption(image, length="normal")["caption"]
    print(f"Opis: {caption}")

    query = model.query(image, "What is the most prominent object in the image?")["answer"]
    print(f"Pytanie: {query}")
```
